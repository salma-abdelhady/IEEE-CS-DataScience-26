### Step 4: Handle Missing Values

##### Use `.replace()` to convert "UNKNOWN" and "ERROR" to np.nan

```python
cafe_sales.replace(["UNKNOWN", "ERROR"], np.nan, inplace=True)
```

### Step 5:Text Sanitization

##### remove hidden spaces from all string columns

```python
for col in cafe_sales.select_dtypes(include='object').columns:
    cafe_sales[col] = cafe_sales[col].str.strip()
```

##### convert all Item names to lowercase for consistency

```python
cafe_sales['Item'] = cafe_sales['Item'].str.lower()
```

```python
cafe_sales['Item']
```

##### use a dictionary to fix any typos in item names if they exist

```python
# array(['Coffee', 'Cake', 'Cookie', 'Salad', 'Smoothie', nan, 'Sandwich', 'Juice', 'Tea'], dtype=object)
```

### Step 6: Fix Data Types
##### fix incorrect data types (Quantity, Price Per Unit, Total Spent .....)

```python
cafe_sales['Quantity'] = pd.to_numeric(cafe_sales['Quantity'])
```

```python
cafe_sales['Price Per Unit'] = pd.to_numeric(cafe_sales['Price Per Unit'])
```

```python
cafe_sales['Total Spent'] = pd.to_numeric(cafe_sales['Total Spent'])
```

```python
cafe_sales.info()
```

##### identify and fix rows where Quantity might be zero or negative

```python
cafe_sales.loc[cafe_sales['Quantity'] <= 0, 'Quantity'] = np.nan
```

```python
cafe_sales['Quantity']
```

### Step 7: Handle Missing Values (Part 2: Fill NaNs)

##### fill missing `Price Per Unit`

```python
cafe_sales['Price Per Unit']
```

```python
cafe_sales['Price Per Unit'].fillna(cafe_sales['Price Per Unit'].median(), inplace=True)
```

```python
cafe_sales['Price Per Unit']
```

##### handle other missing values appropriately (like Payment Method or Location)

```python
cafe_sales['Payment Method'].fillna('unknown', inplace=True)
```

```python
cafe_sales['Location'].fillna('unknown', inplace=True)
```

### Step 8: Logical Validation

##### create a new column `Expected_Total` by multiplying `Quantity * Price Per Unit`

```python
cafe_sales['Expected_Total'] = cafe_sales['Quantity'] * cafe_sales['Price Per Unit']
```

```python
cafe_sales['Expected_Total']
```

##### filter rows where Total Spent is NOT equal to Expected_Total

```python
filtered = cafe_sales[cafe_sales['Total Spent'] != cafe_sales['Expected_Total']]
```

```python
filtered
```

##### fill missing or incorrect Total Spent values using the Expected_Total calculation.

```python
cafe_sales['Total Spent'] = cafe_sales['Expected_Total']
```

### Step 9: Handle Dates

##### convert Transaction Date to datetime

```python
cafe_sales['Transaction Date'] = pd.to_datetime(cafe_sales['Transaction Date'])
```

```python
cafe_sales.info()
```

##### identify transactions with dates in the future (if any) and handle them

```python
from datetime import datetime

today = pd.Timestamp(datetime.today())
cafe_sales.loc[cafe_sales['Transaction Date'] > today, 'Transaction Date'] = np.nan
```

##### extract "Year" and "Month" into separate columns for analysis

```python
cafe_sales['Year'] = cafe_sales['Transaction Date'].dt.year
cafe_sales['Month'] = cafe_sales['Transaction Date'].dt.month
```

```python
cafe_sales['Year']
```

```python
cafe_sales['Month']
```
