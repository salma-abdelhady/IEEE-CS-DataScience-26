Broadcasting: allows Numpy to perform opeartions on arrays with different shapes by virtually expanding dimensions

Rules:

1. The dimensions have the same size

2. One of the dimensions has a size of 1

```python
arr1 = np.array([[1,2,3,4]])
arr2 = np.array([[1],[2],[3],[4]])
```

```python
arr1.shape
```

```python
arr2.shape
```

```python
multi = arr1 * arr2
multi
```
