Problem4:

```python
arr5 = np.random.randint(0,101,(5,5))
print(f"Original Array:\n {arr5}")
# max in each row:
print(f"Max: {arr5.max(axis=1)}")
# sum of each column:
print(f"Sum: {arr5.sum(axis=0)}")
# mean of the matrix:
print(f"Mean: {arr5.mean()}")
# new matrix:
arr6 = (arr5 > 50).astype(int)
print(f"After Masking:\n {arr6}")
```

```python
scores = np.array([45, 88, 92, 30, 50, 76, 99, 12, 65])
# 1- Use Boolean Indexing to filter and print only the passing grades (>= 50)
passed = scores[scores>=50]
print(f"Passed {passed}")

# 2- check if any student got a perfect score of 100
print(np.any(scores == 100))

# 3- check if all students passed
print(len(passed) == len(scores))
```
