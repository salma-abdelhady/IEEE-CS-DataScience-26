Quick Review on Plotting Types


1. Line Plot → shows change over time (trends, time series)
2. Bar Chart → compares categories (ranking, groups)
3. Histogram → shows distribution of one variable (frequency)
4. Scatter Plot → shows relationship between two numeric
variables (correlation, patterns)
5. Box Plot → shows spread, median, and outliers
6. Pie Chart → shows percentage/part of whole

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

Line plot

```python
x = np.array([12, 45, 7, 32, 89, 54, 23, 67, 14, 91])
y = np.array([99, 31, 72, 56, 19, 88, 43, 61, 35, 77])
plt.plot(x, y)
plt.show()
```

Bar chart

```python
plt.bar(x, y)
plt.show()
```
